# 1. Data Types

Every value in Python has a **type**. These are the ones you'll see most:

| Type | What it is | Examples |
|---|---|---|
| `int` | whole number | `7`, `-3`, `0` |
| `float` | decimal number | `3.14`, `-0.5`, `2.0` |
| `str` | text (a "string") | `"hello"`, `'hi'` |
| `bool` | true or false | `True`, `False` |

Use `type()` to check what something is.

````python
age = 20
price = 4.99
name = "Benji"
is_student = True

print(type(age))
print(type(price))
print(type(name))
print(type(is_student))
````

### Math with numbers

````python
print(7 + 2)    # add
print(7 - 2)    # subtract
print(7 * 2)    # multiply
print(7 / 2)    # divide (always gives a float)
print(7 // 2)   # floor divide: drops the decimal part
print(7 % 2)    # remainder (modulo)
print(7 ** 2)   # exponent
````

### Strings

Strings work a lot like lists: they have a length and you can grab characters by index.

````python
word = "hello"
print(len(word))
print(word[0])
print(word[-1])
print(word + " world")   # + joins strings together
print("ha" * 3)          # * repeats them
````

### Converting between types

`int()`, `float()`, and `str()` convert a value to that type.

````python
print(int("5") + 1)
print(float("3.5"))
print(str(5) + "5")
print(int(3.9))     # chops off the decimal, doesn't round
````

````python
"5" + 5
````

Python won't guess for you. Convert one side first, like `int("5") + 5` or `"5" + str(5)`.

This matters because `input()` (which we'll use later) **always** gives back a string, so you'll need `int()` or `float()` if you want to do math with what the user types.

# 2. Lists (Python's version of arrays)

A list holds an **ordered** collection of items. Python lists do the job that arrays do in languages like Java or C, and they can grow and shrink as needed.

Each item has an **index**, and indexes start at **0**.

````python
fruits = ["apple", "banana", "cherry"]
print(fruits)
print(len(fruits))   # how many items?
````

### Indexing

````python
print(fruits[0])    # first item
print(fruits[2])    # third item
print(fruits[-1])   # negative numbers count from the end: -1 is the last item
````

#### What do you think will happen if we ask for an index that doesn't exist?

````python
fruits[3]
````

### Slicing

`list[start:stop]` gives you a piece of the list. It includes `start` but **not** `stop`.

````python
nums = [10, 20, 30, 40, 50]
print(nums[1:4])   # indexes 1, 2, 3
print(nums[:2])    # from the beginning up to (not including) index 2
print(nums[2:])    # from index 2 to the end
````

### Changing a list

````python
fruits[1] = "blueberry"   # replace an item
print(fruits)
````

````python
fruits.append("date")        # add to the end
print(fruits)
fruits.insert(0, "avocado")  # add at a specific index
print(fruits)
````

````python
fruits.remove("cherry")   # remove by value
print(fruits)
last = fruits.pop()       # remove AND return the last item
print(last)
print(fruits)
````

````python
print("apple" in fruits)    # is it in the list?
print("kiwi" in fruits)
print(sorted([5, 2, 9, 1]))  # a new, sorted copy
````

# 3. Conditionals: if / elif / else

So far our code runs every line, top to bottom. **Conditionals** let your program make decisions: run some code only *if* something is true.

### Comparison operators

| Operator | Meaning |
|---|---|
| `==` | equal to |
| `!=` | not equal to |
| `<` / `>` | less than / greater than |
| `<=` / `>=` | less than or equal / greater than or equal |

Careful: `=` **assigns** a value (`x = 5`), while `==` **compares** (`x == 5`).

````python
print(5 == 5)
print(5 != 5)
print(3 < 10)
print("apple" == "apple")
````

### `if`

The indented code only runs when the condition is `True`. Don't forget the colon and the indentation!

````python
age = 20
if age >= 18:
    print("You're an adult")
print("This line always runs (it's not indented)")
````

Try changing `age` to 15 and run it again.

### `else`

`else` runs when the `if` condition is `False`.

````python
temp = 45
if temp > 70:
    print("It's warm")
else:
    print("It's cold")
````

### `elif`

Short for "else if". Use it when you have more than two options. Python checks from the top and **stops at the first one that's true**.

````python
score = 85
if score >= 90:
    print("A")
elif score >= 80:
    print("B")
elif score >= 70:
    print("C")
else:
    print("Keep practicing")
````

Notice that 85 is also `>= 70`, but we only got `B`, because Python stopped at the first true condition. That means **order matters**.

### Combining conditions with `and`, `or`, `not`

````python
age = 20
has_ticket = True

if age >= 18 and has_ticket:
    print("Come on in")

if age < 13 or age > 65:
    print("Discount price")
else:
    print("Regular price")

print(not has_ticket)
````

### Conditionals with lists

The `in` keyword from earlier works great inside an `if`.

````python
fruits = ["apple", "banana", "cherry"]
if "kiwi" in fruits:
    print("We have kiwi")
else:
    print("No kiwi today")

if len(fruits) == 0:
    print("The list is empty")
else:
    print("The list has", len(fruits), "items")
````

# 4. Loops

Loops let you repeat code without copy-pasting it.

### The `for` loop

"For each item in this list, do something."

````python
fruits = ["apple", "banana", "cherry"]
for fruit in fruits:
    print(fruit)
````

### `range()`

`range(n)` counts from 0 up to (but not including) `n`. You can also give it a start and a step.

````python
for i in range(5):
    print(i)
````

````python
print(list(range(2, 6)))      # start at 2, stop before 6
print(list(range(0, 10, 2)))  # count by 2s
````

### Looping with an index

Sometimes you need the position too. `range(len(list))` gives you every valid index.

````python
for i in range(len(fruits)):
    print(i, fruits[i])
````

### Pattern: keeping a running total

````python
nums = [10, 20, 30, 40, 50]
total = 0
for n in nums:
    total += n     # same as total = total + n
print(total)
````

### The `while` loop

Keeps going **as long as** a condition is true. Make sure something inside the loop eventually makes it false, or it will run forever!

````python
count = 0
while count < 3:
    print("count is", count)
    count += 1
````

### `break` and `continue`

`break` exits the loop immediately. `continue` skips to the next round.

````python
for n in range(10):
    if n == 5:
        break
    print(n)
````

````python
for n in range(10):
    if n % 2 == 0:
        continue    # skip even numbers
    print(n)
````

# 5. Dictionaries

A dictionary stores **key -> value** pairs. Instead of looking things up by position (like a list), you look them up by key.

````python
individual = {
    "name": "joe",
    "pet": "dog",
    "age": 13
}
print(individual["name"])
````

````python
individual.items()
````

````python
individual.keys()
````

````python
individual.values()
````

### Adding, changing, and removing

````python
individual["job"] = "teacher"   # new key
individual["age"] = 14          # existing key gets a new value
print(individual)
````

````python
del individual["job"]
individual
````

### .pop(key, default)

**key**: the key to remove

**default**: what to return if the key isn't found

````python
individual["job"] = "teacher"
print(individual.pop("job", None))             # removes it and returns the value
print(individual.pop("social security", None)) # returns None since the key doesn't exist
````

#### Now, try deleting the key "social security" with `del`. What do you think will happen?

````python
del individual["social security"]
````

#### We can use `try`/`except` to handle errors without crashing.

````python
try:
    individual.pop("social security")
except KeyError:
    print("social security isn't a key")
````

### Values can be lists (and loops work on dictionaries too)

````python
individual["pet"] = ["cat", "dog", "fish"]
print(individual["pet"][0])
````

````python
for key in individual:                       # loops over the keys
    print(key, "->", individual[key])
````

````python
for val in individual.values():                       # loops over the values
    print(val)
````

````python
for key, value in individual.items():        # loops over key AND value
    print(key, "->", value)
````

# 6. Sets

Sets are like lists, but they are <b><i>UNORDERED</i></b> (so they have <b><i>no index</i></b>) and every item appears <b><i>only once</i></b>.

````python
set1 = {"hello", "goodbye", "ciao"}
set1
````

#### Can we grab the item at index 1 like with a list?

````python
set1[1]
````

````python
for item in set1:
    print(item)
````

````python
set1.add("hola")
set1.remove("ciao")
print(set1)
print("hello" in set1)
````

````python
animals1 = {"dog", "cat", "goat", "otter"}
animals2 = {"fish", "whale", "shark", "otter"}
print(animals1.intersection(animals2))   # in both
print(animals1.union(animals2))          # in either
````

#### Watch out: what type is `mystery`?

````python
mystery = {}
empty_set = set()

print(type(mystery))
print(type(empty_set))
````

A neat trick: turning a list into a set removes the duplicates.

````python
print(set([1, 2, 2, 3, 3, 3]))
````

# 7. Handy Extras

## print vs. return

`print` shows something on the screen. `return` hands a value back so the rest of your program can use it.

````python
def add_return(a, b):
    return a + b

def add_print(a, b):
    print(a + b)

x = add_return(2, 3)   # nothing is shown, but x holds 5
y = add_print(2, 3)    # prints 5, but y holds nothing useful

print(x)
print(y)
````

## f-strings

f-strings let you drop variables straight into a string. Put an `f` before the quotes and wrap variables in `{}`.

````python
dog = "Barky"
owner = "Joe"
print(owner + " has a dog named " + dog)   # the old way
print(f"{owner} has a dog named {dog}")    # the f-string way
````

# 8. Bonus: Classes (optional)

A class is a **blueprint** for making objects. Each object (an "instance") gets its own copy of the data.

````python
class MyClass:
    def __init__(self, name, year):
        self.name = name
        self.year = year

    def statement(self):
        print("My name is " + self.name)

    def my_year(self):
        print("My year is " + str(self.year))
````

What we just saw:

- `class MyClass` creates the blueprint.
- `__init__` runs every time a new instance is made. Here it needs a `name` and a `year`.
- `self` means "this particular instance". It's how the object accesses its own data.
- `statement` and `my_year` are **methods**: functions that belong to the class.

````python
dina = MyClass("Dina", "3rd")
dina.statement()

preston = MyClass("Preston", "3rd")
preston.statement()
preston.my_year()
````

# Recap

- **Data types**: `int`, `float`, `str`, `bool`; check with `type()`, convert with `int()` / `float()` / `str()`
- **Lists**: ordered, indexed from 0; `append`, `insert`, `remove`, `pop`, slicing, `in`
- **Conditionals**: `if` / `elif` / `else` to make decisions; `==`, `!=`, `<`, `>`, `and`, `or`, `not`
- **Loops**: `for` to go through items, `range()` to count, `while` to repeat until a condition changes, `break`/`continue`
- **Dictionaries**: key -> value; loop with `.items()`; `.pop(key, default)` and `try`/`except` for missing keys
- **Sets**: unordered, no duplicates, great for removing repeats
- **Extras**: `return` vs. `print`, shorter functions, f-strings
