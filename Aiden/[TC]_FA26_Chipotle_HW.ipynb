### Step 9. Which was the most-ordered item?

```python
# Hint: When sorting values, you can pass ascending = False
most_ordered = (chipo.groupby("item_name")["quantity"].sum().sort_values(ascending=False))
most_ordered.index[0]
```

### Step 10. For the most-ordered item, how many items were ordered?

```python
chipo.groupby("item_name")["quantity"].sum().max()
```

### Step 11. What was the most ordered item in the choice_description column?

```python
most_ordered_choice = (chipo.groupby("choice_description")["quantity"].sum().sort_values(ascending=False))
most_ordered_choice.index[0]
```

### Step 12. For the most-ordered item in choice_description, how many items were ordered?

```python
chipo.groupby("choice_description")["quantity"].sum().max()
```
